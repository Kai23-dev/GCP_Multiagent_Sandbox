## Flatten Taxonomy

The section the L1-L3 taxonomy into a pandas DataFrame, transforms it by renaming columns, dropping duplicates, extracting GL account codes and descriptions, and adding placeholder columns for flags and dates.

```python
# prompt: read the contents of /content/L1-L3_Taxonomy.csv and print out. Use pandas
df = pd.read_csv('/content/L1-L3_Taxonomy.csv')
df
```

```python
df_transformed = df.iloc[:, :4].copy()
print (df_transformed.columns)
```

```python
df_transformed.rename(columns={
    'Sourcing Lane': 'Sourcing_Lane',
    'L1': 'L1_Category',
    'L2': 'L2_Category',
    'GL Account (L3)': 'L3_Category',
}, inplace=True)
```

```python
# prompt: remove duplicates in df_transformed
df_transformed.drop_duplicates(inplace=True)
```

```python
# prompt: Create new cols GL_Account_L3_Code and GL_Account_Description from df_transformed["L3_Category"]  which looks something like 6616 - CRRT Supply Expense. So four digits (the account code), space character(s), dash, space character(s) and then the description.
df_transformed[['GL_Account_L3_Code', 'GL_Account_Description']] = df_transformed['L3_Category'].str.extract(r'(\d{4})\s*-\s*(.*)')
df_transformed.head()
```

```python
df_transformed['Is_Direct_Flag'] = df_transformed['Sourcing_Lane'].apply(lambda x: True if x == 'Direct' else False)
df_transformed['Category_Owner'] = '' # Placeholder, assuming this data is not in the original CSV
df_transformed['Effective_Start_Date'] = pd.to_datetime('today').date() # Placeholder
df_transformed['Effective_End_Date'] = None # Placeholder
```

```python
df_transformed.head()
```

```python
df_transformed.to_csv('/content/L1-L3_SSI_Agent_Taxonomy.csv', index=False)
```

## Build out iProc Taxonomy
L4s and joined against L1-L3s

```python
import pandas as pd
iproc = pd.read_csv('/content/iProc_Raw_L4s.csv')
iproc.head()
```

```python
# prompt: separate the "(d\{4})" from List col header in iproc df. Rename List col name to raw. The regex should go into a new col named account while the part before should go into a new col "L4_Category"

iproc[['L4_Category', 'GL_Account_L3_Code']] = iproc['List'].str.extract(r'(.*)\s*\((\d{4})\)')
iproc['L4_Category'] = iproc['L4_Category'].apply(lambda x: x.title())
iproc.rename(columns={'List': 'raw'}, inplace=True)
print(iproc.head())
```

```python
# prompt: show frequency distribution for iproc['GL_Account_L3_Code'] and sort by most occurring to least occurring. concat the L4_Category names in a new col

iproc_counts = iproc['GL_Account_L3_Code'].value_counts().reset_index()
iproc_counts.columns = ['GL_Account_L3_Code', 'frequency']
iproc_counts = iproc_counts.sort_values(by='frequency', ascending=False)

l4_categories = iproc.groupby('GL_Account_L3_Code')['L4_Category'].apply(lambda x: ', '.join(x)).reset_index()
l4_categories.columns = ['GL_Account_L3_Code', 'L4_Categories']

iproc_counts = pd.merge(iproc_counts, l4_categories, on='GL_Account_L3_Code', how='left')
```

```python
# prompt: Add the L1s and L2s to iproc_counts by joining to df_transformed using GL_Account_L3_Code. Move 'L1_Category', 'L2_Category' cols to the beginning of the df

iproc_counts = pd.merge(iproc_counts, df_transformed[['GL_Account_L3_Code', 'L1_Category', 'L2_Category']], on='GL_Account_L3_Code', how='left')
cols = ['L1_Category', 'L2_Category'] + [col for col in iproc_counts.columns if col not in ['L1_Category', 'L2_Category']]
iproc_counts = iproc_counts[cols]
```

# Mary's Taxonomy
most recent code to create sco-kb-dev-1a9n.sco_supplier_classification.SPEND_TAXONOMY

```python
# prompt: Read /content/Full_Hierarchy_03022026.csv into a pandas df

import pandas as pd

mary_df = pd.read_csv('/content/Full_Hierarchy_03022026.csv')
print (mary_df.shape)
mary_df.tail()
```

```python
# prompt: filter out into a new df mary_df_clnd where col Change Type != Remove and does not start with "Becomes"
# prompt: drop PRISM Commodity Level 4, PRISM Commodity Level 5, iProc Product Category, Financial Class, Special Reporting, Purchase Type, Change Type from mary_df_clnd

mary_df_clnd = mary_df[~mary_df['Change Type'].isin(['Remove'])]
mary_df_clnd = mary_df_clnd[~mary_df_clnd['Change Type'].astype(str).str.startswith('Becomes')]
mary_df_clnd.shape
```

```python
# prompt: In mary_df_clnd, remove the string pattern (\d{4}) including the parenthesis from PRISM Commodity Name, and unique that list

mary_df_clnd['PRISM Commodity Level 3'] = mary_df_clnd['PRISM Commodity Level 3'].str.replace(r'(\(\d{4}\))', '', regex=True)
mary_df_clnd['PRISM Commodity Level 3'] = mary_df_clnd['PRISM Commodity Level 3'].str.strip()
mary_df_clnd['PRISM Commodity Name'] = mary_df_clnd['PRISM Commodity Name'].str.replace(r'(\(\d{4}\))', '', regex=True)
mary_df_clnd['PRISM Commodity Name'] = mary_df_clnd['PRISM Commodity Name'].str.strip()
mary_df_clnd["Account Description"] = mary_df_clnd["Account.1"].str.split(" - ", n=1).str[1]
mary_df_clnd = mary_df_clnd.drop(columns=['PRISM Commodity Level 4', 'PRISM Commodity Level 5', 'iProc Product Category', 'Account.1', 'Financial Class', 'Special Reporting', 'Purchase Type', 'Change Type'])
mary_df_clnd.tail()
```

```python
unique_prism_names = mary_df_clnd['PRISM Commodity Name'].unique()
unique_prism_names.sort()
print (unique_prism_names.shape)
unique_prism_names[0:5]
```

```python
# prompt: write unique_prism_names to txt file and mary_df_clnd to csv

with open('/content/out/unique_prism_names.txt', 'w') as f:
  for name in unique_prism_names:
    f.write(f"{name}\n")

mary_df_clnd.to_csv('/content/out/L1-L4_SSI_Agent_Taxonomy.csv', index=False)
```
